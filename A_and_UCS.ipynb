```python
import heapq

# A* Search Algorithm
def astar(graph, heuristic, start, goal):

    # (f = g + h, g, current node, path)
    pq = [(heuristic[start], 0, start, [])]

    # To keep track of visited nodes
    visited = set()

    # Continue until the priority queue becomes empty
    while pq:

        # Remove the node with the smallest f value
        f, g, node, path = heapq.heappop(pq)

        # Skip if node is already visited
        if node in visited:
            continue

        # Mark the node as visited
        visited.add(node)

        # Add current node to the path
        path = path + [node]

        # Check if goal is reached
        if node == goal:
            return path, g

        # Explore all neighboring nodes
        for neighbor, cost in graph[node]:

            # Process only unvisited neighbors
            if neighbor not in visited:

                # Calculate actual cost from start
                new_g = g + cost

                # Calculate estimated total cost
                new_f = new_g + heuristic[neighbor]

                # Add neighbor to the priority queue
                heapq.heappush(pq, (new_f, new_g, neighbor, path))

    # Return None if no path exists
    return None


# -------------------- Graph --------------------
graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2), ('E', 5)],
    'C': [('F', 3)],
    'D': [('G', 1)],
    'E': [('G', 2)],
    'F': [('G', 2)],
    'G': []
}

# ---------------- Heuristic Values ----------------
heuristic = {
    'A': 4,
    'B': 3,
    'C': 4,
    'D': 1,
    'E': 2,
    'F': 2,
    'G': 0
}

start = 'A'
goal = 'G'

result = astar(graph, heuristic, start, goal)

if result:
    print("Shortest Path:", result[0])
    print("Total Cost:", result[1])
else:
    print("No path found.")
```

```python
import heapq

# Uniform Cost Search (UCS) Algorithm
def ucs(graph, start, goal):

    # Priority Queue stores:
    # (total_cost, current_node, path)
    pq = [(0, start, [])]

    # To keep track of visited nodes
    visited = set()

    # Continue until the priority queue becomes empty
    while pq:

        # Remove the node with the smallest cost
        cost, node, path = heapq.heappop(pq)

        # Skip if the node has already been visited
        if node in visited:
            continue

        # Mark the node as visited
        visited.add(node)

        # Add the current node to the path
        path = path + [node]

        # Check if the goal node is reached
        if node == goal:
            return path, cost

        # Explore all neighboring nodes
        for neighbor, weight in graph[node]:

            # Process only unvisited neighbors
            if neighbor not in visited:

                # Add neighbor to the priority queue
                # New cost = Current cost + Edge weight
                heapq.heappush(pq, (cost + weight, neighbor, path))

    # Return None if no path exists
    return None


# ---------------- Graph ----------------
graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2), ('E', 5)],
    'C': [('F', 3)],
    'D': [('G', 1)],
    'E': [('G', 2)],
    'F': [('G', 2)],
    'G': []
}

# ---------------- Driver Code ----------------
start = 'A'
goal = 'G'

result = ucs(graph, start, goal)

if result:
    print("Shortest Path:", result[0])
    print("Total Cost:", result[1])
else:
    print("No path found.")
```
